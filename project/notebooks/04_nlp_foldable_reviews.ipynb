# 04 — NLP: Foldable Review Sentiment + Topic Modelling

Filter reviews to foldable models via the `is_foldable` mapping (record corpus size
honestly). Pipeline: lowercase, stopword removal, lemmatization, tokenization.
VADER sentiment (overall / by model / price tier / country); LDA topics on the
complaint subset (5-8 topics, manually labelled). Outputs: ranked pain points.
